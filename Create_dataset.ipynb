```python
import random

def generate_random_name(syllable_count):
    vowels = 'aeiou'
    consonants = 'bcdfghjklmnpqrstvwxyz'
    syllables = [random.choice(consonants) + random.choice(vowels) for _ in range(syllable_count)]
    return ''.join(syllables).capitalize()


def Tamil_marks():
  ls = [ random.randint(1,100) for _ in range(1,151)]
  return ls

def English_marks():
  ls = [ random.randint(1,100) for _ in range(1,151)]
  return ls

def Maths_marks():
  ls = [ random.randint(1,100) for _ in range(1,151)]
  return ls

def Science_marks():
  ls = [ random.randint(1,100) for _ in range(1,151)]
  return ls

def SocialScince_Marks():
  ls = [ random.randint(1,100) for _ in range(1,151)]
  return ls

def Get_name():
   ls = [(generate_random_name(syllable_count = 2)) for _ in range(1,151)]
   return ls


def Name_with_marks():
  data = []
  for i in range(0,150):
    name_with_marks = []
    name_with_marks.append(Get_name()[i])
    name_with_marks.append(Tamil_marks()[i])
    name_with_marks.append(English_marks()[i])
    name_with_marks.append(Maths_marks()[i])
    name_with_marks.append(Science_marks()[i])
    name_with_marks.append(SocialScince_Marks()[i])
    data.append(name_with_marks)
  return data
```

```python


def list_to_csv(data_list, file_name = 'marks.csv'):
    with open(file_name, 'w', newline='') as csvfile:
        csv_writer = csv.writer(csvfile)
        csv_writer.writerows(data_list)

list_to_csv(Name_with_marks())
print("Successfully Mark")


```

```python
import numpy as np
import pandas as pd

def generate_synthetic_dataset(num_samples=100, num_features=5):
    # Generate random features (between 0 and 1)
    X = np.random.rand(num_samples, num_features)

    # Generate random labels (0 or 1)
    y = np.random.randint(2, size=num_samples)

    # Create a pandas DataFrame for convenience (optional)
    feature_names = [f"feature_{i}" for i in range(num_features)]
    df = pd.DataFrame(X, columns=feature_names)
    df['target'] = y

    return df

if __name__ == "__main__":
    # Specify the dataset parameters
    num_samples = 100
    num_features = 5

    # Generate the synthetic dataset
    dataset = generate_synthetic_dataset(num_samples, num_features)

    # Display the first few rows of the dataset
    print(dataset.head())
```
