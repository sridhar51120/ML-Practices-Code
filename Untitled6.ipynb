```python
# Suppose you have a dataset with a "Fruit" column containing categorical values: ["Apple", "Orange", "Banana"].
# After one-hot encoding, the "Fruit" column will be transformed into three separate binary columns: ["Fruit_Apple", "Fruit_Orange", "Fruit_Banana"], where each column represents the presence of a specific fruit.

# Label Encoding:
# Label encoding is used to convert categorical values into numeric labels. Each category is assigned a unique integer value.
# Example:
# Consider the same "Fruit" column as above. After label encoding, the "Fruit" column will be represented as [0, 1, 2] for ["Apple", "Orange", "Banana"].

# Ordinal Encoding:
# Ordinal encoding is similar to label encoding, but it is used when there is a natural order or ranking among the categories.
# Example:
# Suppose you have a "Size" column with values ["Small", "Medium", "Large"]. Ordinal encoding will convert it to [0, 1, 2] to reflect the order of sizes.

# Binary Encoding:
# Binary encoding combines one-hot encoding and label encoding to represent categorical values with binary digits. It first encodes the categories as integers and then converts the integers to binary form.
# Example:
# Using the same "Fruit" column, binary encoding could produce [0, 1, 10] for ["Apple", "Orange", "Banana"], where each binary digit represents the presence of a specific fruit.

# Target Encoding:
# Target encoding uses the target variable to encode categorical data. It replaces each category with the mean (or other statistic) of the target variable for that category.
# Example:
# Suppose you have a "City" column and you want to predict house prices. Target encoding will replace each city name with the mean house price in that city.
```

```python
from sklearn.preprocessing import LabelEncoder
# Label Encoding Meethod --> method to both fit the encoder to the data and transform the categories into numerical labels
label_encoder = LabelEncoder()

encoded_labels = label_encoder.fit_transform(coloumn)

print(f"Encoded Labels :: {encoded_labels}")
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

from sklearn.datasets import load_iris
iris = load_iris()
data = iris.data
feature_names = iris.feature_names

df = pd.DataFrame(data, columns=feature_names)

# MinMaxScaler object
scaler = MinMaxScaler()

# Apply Min-Max scaling to the numerical features
scaled_data = scaler.fit_transform(df)

# Convert the scaled_data array back to a DataFrame
scaled_df = pd.DataFrame(scaled_data, columns=feature_names)

# Print the scaled data
print("\nScaled Data:")
print(scaled_df.head())
```
