```python
import math
data = [
    {'Outlook': 'Sunny', 'Temperature': 'Hot', 'Humidity': 'High', 'Windy': False, 'PlayGolf': 'No'},
    {'Outlook': 'Sunny', 'Temperature': 'Hot', 'Humidity': 'High', 'Windy': True, 'PlayGolf': 'No'},
    {'Outlook': 'Overcast', 'Temperature': 'Hot', 'Humidity': 'High', 'Windy': False, 'PlayGolf': 'Yes'},
    {'Outlook': 'Rainy', 'Temperature': 'Mild', 'Humidity': 'High', 'Windy': False, 'PlayGolf': 'Yes'},
    {'Outlook': 'Rainy', 'Temperature': 'Cool', 'Humidity': 'Normal', 'Windy': False, 'PlayGolf': 'Yes'},
    {'Outlook': 'Rainy', 'Temperature': 'Cool', 'Humidity': 'Normal', 'Windy': True, 'PlayGolf': 'No'},
    {'Outlook': 'Overcast', 'Temperature': 'Cool', 'Humidity': 'Normal', 'Windy': True, 'PlayGolf': 'Yes'},
    {'Outlook': 'Sunny', 'Temperature': 'Mild', 'Humidity': 'High', 'Windy': False, 'PlayGolf': 'No'},
    {'Outlook': 'Sunny', 'Temperature': 'Cool', 'Humidity': 'Normal', 'Windy': False, 'PlayGolf': 'Yes'},
    {'Outlook': 'Rainy', 'Temperature': 'Mild', 'Humidity': 'Normal', 'Windy': False, 'PlayGolf': 'Yes'},
    {'Outlook': 'Sunny', 'Temperature': 'Mild', 'Humidity': 'Normal', 'Windy': True, 'PlayGolf': 'Yes'},
    {'Outlook': 'Overcast', 'Temperature': 'Mild', 'Humidity': 'High', 'Windy': True, 'PlayGolf': 'Yes'},
    {'Outlook': 'Overcast', 'Temperature': 'Hot', 'Humidity': 'Normal', 'Windy': False, 'PlayGolf': 'Yes'},
    {'Outlook': 'Rainy', 'Temperature': 'Mild', 'Humidity': 'High', 'Windy': True, 'PlayGolf': 'No'}
]
def entropy(dataset):
    label_counts = {}
    total_samples = len(dataset)
    for data_point in dataset:
        label = data_point['PlayGolf']
        if label not in label_counts:
            label_counts[label] = 0
        label_counts[label] += 1
    entropy = 0
    for label in label_counts:
        probability = label_counts[label] / total_samples
        entropy -= probability * math.log2(probability)
    return entropy
def information_gain(dataset, attribute):
    total_entropy = entropy(dataset)
    attribute_entropy = 0
    attribute_value_counts = {}
    for data_point in dataset:
        attribute_value = data_point[attribute]
        if attribute_value not in attribute_value_counts:
            attribute_value_counts[attribute_value] = []
        attribute_value_counts[attribute_value].append(data_point)
    for attribute_value, subset in attribute_value_counts.items():
        probability = len(subset) / len(dataset)
        attribute_entropy += probability * entropy(subset)
    return total_entropy - attribute_entropy
def gain_ratio(dataset, attribute):
    info_gain = information_gain(dataset, attribute)
    split_info = 0
    attribute_value_counts = {}

    for data_point in dataset:
        attribute_value = data_point[attribute]
        if attribute_value not in attribute_value_counts:
            attribute_value_counts[attribute_value] = 0
        attribute_value_counts[attribute_value] += 1
    for attribute_value_count in attribute_value_counts.values():
        probability = attribute_value_count / len(dataset)
        split_info -= probability * math.log2(probability)
    if split_info == 0:
        return 0
    return info_gain / split_info
attributes = data[0].keys()
max_info_gain_attribute = max(attributes, key=lambda attr: information_gain(data, attr))
max_gain_ratio_attribute = max(attributes, key=lambda attr: gain_ratio(data, attr))

print("Attribute with Maximum Information Gain:", max_info_gain_attribute)
print("Attribute with Maximum Gain Ratio:", max_gain_ratio_attribute)
```
