```python
# Creating A DataFrame from an Existing DataFrame

old_df = pd.DataFrame({'team': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'],
                       'points': [18, 22, 19, 14, 14, 11, 20, 28],
                       'assists': [5, 7, 7, 9, 12, 9, 9, 4],
                       'rebounds': [11, 8, 10, 6, 6, 7, 9, 12]})

# print(old_df)
new_df = old_df[['rebounds','assists']].copy()
print(new_df)
```

```python
# Getting A Column of the DataFrame

print("Column of the DataFrame\n")
col = new_df.columns
print(col)
```

```python
# Adding Column

data = {'Name': ['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Height': [5.1, 6.2, 5.1, 5.2],
        'Qualification': ['Msc', 'MA', 'Msc', 'Msc']}

result = ['Fail','Fail','Fail','Pass']

dataFrame = pd.DataFrame(data)


print("Old DataFrame\n")
print(dataFrame)
print("\nNew DataFrame\n")

dataFrame['result'] = result

print(dataFrame)
```

```python
# Column Selection in Pandas

# print(data)
print("Coloumn Selection\n")
print(dataFrame['Name'])
```

```python
# Add Rows in DataFrame
dict = {'Name':['Martha', 'Tim', 'Rob', 'Georgia'],
		'Maths':[87, 91, 97, 95],
		'Science':[83, 99, 84, 76]
	}

df = pd.DataFrame(dict)
print("Old DataFrame\n")
print(df)

df.loc[len(df.index)] = ['Amy', 89, 93]

print("\n",df,"\n")
```

```python
# Deletion of Row

print("Old DataFrame\n")

print(df)
print("\n")

row_del = df.drop(1) # or row_del = df.drop(label = 1) --> Refers Index Value

print(row_del)
```

```python
# Create two different dataframes and concatenate.(Union operations of tables).

# DataFrame --> 01

data1 = {'product': ['computer', 'tablet', 'monitor', 'printer'],
         'brand': ['AA', 'BB', 'CC', 'DD'],
         'price': [1200, 350, 500, 150]
         }

df1 = pd.DataFrame(data1)
print("\nDataFrame 01 \n")
print(df1)


# DataFrame -->  02

data2 = {'product': ['keyboard', 'mouse', 'speakers', 'scanner'],
         'brand': ['EE', 'FF', 'GG', 'HH'],
         'price': [120, 50, 200, 180]
         }

df2 = pd.DataFrame(data2)
print("\n DataFrame 02 \n")
print(df2)


#concat using np.concat([...])

union = pd.concat([df1, df2])
print("\n New DataFrame \n")
print(union)
```

```python
# Merge Two DataFrames

data1 = {
  "name": ["Sally", "Mary", "John"],
  "age": [50, 40, 30]
}

data2 = {
  "name": ["Sally", "Peter", "Micky"],
  "age": [77, 44, 22]
}

df1 = pd.DataFrame(data1)
print("DataFrame 01\n")
print(df1,"\n")

df2 = pd.DataFrame(data2)
print("\nDataFrame 02 \n")
print(df2,"\n")

newdf = df1.merge(df2, how='right')
print("New Merged DataFrame\n")
print(newdf)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Import Done")
```

```python
# Lines

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle = 'dotted')

"""
1) 'solid' (default)	'-'
2)  'dotted'	':'
3)  'dashed'	'--'
4)  'dashdot'	'-.'

"""

plt.show()
```

```python
ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle = 'dashed')
plt.show()
```

```python
ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints)  # default = 'solid'
plt.show()
```

```python
# Pie Charts

ypoints = np.array([35, 25, 25, 15])
label = ['Tamil','Maths','Physics','Chemistry']
Colors = ['r','green','skyblue','blue']
"""
Colors -->
    'r' - Red
    'g' - Green
    'b' - Blue
    'c' - Cyan
    'm' - Magenta
    'y' - Yellow
    'k' - Black
    'w' - White

"""
plt.pie(ypoints,labels=label,colors = Colors)
plt.legend(title = "Results :: ")
print("Pie Chart  :: ")
plt.show()
```

```python
# Bar Chart

ypoints = np.array([35, 25, 25, 15])
xpoints = ['Tamil','Maths','Physics','Chemistry']

plt.bar(xpoints,ypoints)
print("Bar Chart :: \n")
plt.show()
```

```python
# Histogram

print("Histogram :: \n")

x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

```python
# Scatter Plot

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x, y)
print("Scatter \n")
plt.show()
```
