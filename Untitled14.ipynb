```python
import re
```

```python
compare_value = False
```

```python
file = open('/content/test.txt','r')

text  = []

for i in file:
  text.append(str(i[:]))
file.close()
print("Done")
```

```python
def compare_strings(string1, string2):
    pattern = re.compile(string2)
    match = re.search(pattern, string1)

    if match:
        return True
```

```python
for i in text:
  if compare_strings(i,"asdjksdsad"):
    compare_value = True
```

```python
if compare_value:
  print("Done")
```
