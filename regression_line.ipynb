```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import csv
print("Import Done")
```

```python
data = pd.read_csv('/content/data.csv')
online_booking = data['Online boarding'][0:30]
seatcomfort = data['Seat comfort'][0:30]
print("Done")
```

# **Sum of Products and Sum of Squares**

```python
# Sum of Products
x_bar = 0
y_bar = 0
sum_x = 0
sum_y = 0
for i in online_booking:
  sum_x = sum_x + i

for i in seatcomfort:
  sum_y = sum_y + i
x_bar = sum_x / len(online_booking)
y_bar = sum_y / len(seatcomfort)

x_bar = round(x_bar,2)
y_bar = round(y_bar,2)

print(x_bar)
print(y_bar)
```

```python
sub_x_x_bar = []
sub_y_y_bar = []
for i in online_booking:
  val = i - x_bar
  sub_x_x_bar.append(round(val,4))

for i in seatcomfort:
  val = i - y_bar
  sub_y_y_bar.append(round(val,4))

print(sub_x_x_bar)
print(sub_y_y_bar)
```

```python
product_x_y = []
for i in range(0,30):
  val1 = sub_x_x_bar[i]
  val2 = sub_y_y_bar[i]
  mul = val1 * val2
  product_x_y.append(round(mul,4))

print(product_x_y)
```

```python
square_x = []
for i in range(0,30):
  val1 = sub_x_x_bar[i]
  mul = val1 * val1
  square_x.append(round(mul,4))

print(square_x)
```

```python
sum_of_products  = 0
for i in product_x_y:
  sum_of_products = sum_of_products + i

print(sum_of_products)
```

```python
sum_of_square = 0
for i in square_x:
  sum_of_square = sum_of_square + i

sum_of_square = round(sum_of_square,4)

print(sum_of_square)
```

# **M**

```python
m = 0
m = sum_of_products / sum_of_square
m = round(m,4)
print(m)
```

# **c**

```python
c = 0
c = y_bar - (m * x_bar)
c = round(c,4)
print(c)
```

```python
data = []
for i in range(0,30):
  attr = []
  attr.append(online_booking[i])
  attr.append(seatcomfort[i])
  attr.append(sub_x_x_bar[i])
  attr.append(sub_y_y_bar[i])
  attr.append(product_x_y[i])
  attr.append(square_x[i])
  data.append(attr)

print(data)
```

```python
tittle = ['x_bar','y_bar','x - x_bar','y - y_bar','(x - x_bar) * (y - y_bar)','(x - x_bar) * (x - x_bar)']
print(tittle)
```

```python
df = pd.DataFrame(data,columns=tittle)
print(df)
```

```python
df.to_csv('myData.csv',index=False)
print("Csv File Created")
# csv = pd.read_csv('myData.csv')
# print(csv)
```
